```python
import numpy as np
np.random.seed(42)
diameters = np.random.normal(loc=1.2, scale=0.3, size=200)
diameters = diameters[diameters > 0]
```

```python
mean_d = np.mean(diameters)
std_d = np.std(diameters)
min_d, max_d = np.min(diameters), np.max(diameters)
print(mean_d, std_d, min_d, max_d)
```

```python
import pandas as pd
df = pd.DataFrame({
    'droplet_id': range(1, len(diameters) + 1),
    'diameter_mm': diameters,
    'radius_mm': diameters / 2,
    'area_mm2': np.pi * (diameters / 2) ** 2
})
df.head()
```
