```python
diameters = [1.2, 0.9, 1.5, 2.1, 0.7, 1.8] 
large = [d for d in diameters if d > 1.3] 
squared = [d ** 2 for d in diameters] 
print(large) 
print(squared)
```

```python
print(diameters[0])
print(diameters[-1])
print(diameters[1:4])
print(diameters[:3])
print(diameters[::-1])
```

```python
def droplet_stats(values):
    return min(values), max(values), sum(values) / len(values)

lo, hi, avg = droplet_stats(diameters)
print(f"min={lo}, max={hi}, avg={avg:.2f}")
```

```python
droplet_readings = {
    'sample_A': [1.1, 1.3, 0.9],
    'sample_B': [1.8, 2.0, 1.5],
}
for name, values in droplet_readings.items():
    avg = sum(values) / len(values)
    print(f"{name}: average diameter = {avg:.2f} mm")
```

```python
def safe_average(values):
    if len(values) == 0:
        return None
    return sum(values) / len(values)

print(safe_average([1.0, 2.0, 3.0]))
print(safe_average([]))
```
