```python
class Droplet: 
    def __init__(self, diameter_mm): 
        self.diameter_mm = diameter_mm 
        self.radius_mm = diameter_mm / 2 
        self.area_mm2 = 3.14159 * self.radius_mm ** 2 

    def is_large(self, threshold=1.5): 
        return self.diameter_mm > threshold

    def __repr__(self): 
        return f"Droplet(diameter={self.diameter_mm:.2f}mm)" 

d2 = Droplet(1.8) 
print(d2.is_large()) 
print(d2)
```

```python
droplets =[
    Droplet(1.1), 
    Droplet(1.6), 
    Droplet(0.9), 
    Droplet(2.0), 
    Droplet(1.4)
    ] 
large_ones = [d for d in droplets if d.is_large()] 
print(large_ones) 
print(f"{len(large_ones)} out of {len(droplets)} droplets are large")
```
